```python
from sqlalchemy import create_engine
import pandas as pd
import numpy as np

engine = create_engine(
    "postgresql+psycopg2://logistics:logistics@localhost:5432/logistics"
)

full_df = pd.read_sql(
    "SELECT * FROM training_data",
    engine,
    parse_dates=["pickup_datetime", "dropoff_datetime"],
)

cutoff = pd.Timestamp("2024-03-01")

train_full = full_df[full_df.pickup_datetime < cutoff].copy()
test_full = full_df[full_df.pickup_datetime >= cutoff].copy()

print(train_full.shape)
print(test_full.shape)
```

```python
def add_basic_features(df):
    df = df.copy()

    dt = df["pickup_datetime"]

    df["pickup_hour"] = dt.dt.hour
    df["pickup_day_of_week"] = dt.dt.dayofweek
    df["pickup_month"] = dt.dt.month

    df["is_weekend"] = (df["pickup_day_of_week"] >= 5).astype(np.int8)

    df["rush_hour"] = (
        ((df["pickup_hour"] >= 7) & (df["pickup_hour"] <= 9))
        |
        ((df["pickup_hour"] >= 16) & (df["pickup_hour"] <= 19))
    ).astype(np.int8)

    df["is_night"] = (
        (df["pickup_hour"] >= 22)
        |
        (df["pickup_hour"] <= 5)
    ).astype(np.int8)

    df["passenger_count"] = df["passenger_count"].clip(1, 6)

    df["is_large_group"] = (
        df["passenger_count"] >= 4
    ).astype(np.int8)

    df["log_distance"] = np.log1p(df["trip_distance"])

    return df

train_full = add_basic_features(train_full)
test_full = add_basic_features(test_full)
```

```python
zones = pd.read_csv("../data/raw/taxi_zone_lookup.csv")[
    ["LocationID", "Borough"]
]

def add_boroughs(df, zones):

    df = df.merge(
        zones.rename(
            columns={
                "LocationID": "pickup_location_id",
                "Borough": "pickup_borough",
            }
        ),
        on="pickup_location_id",
        how="left",
    )

    df = df.merge(
        zones.rename(
            columns={
                "LocationID": "dropoff_location_id",
                "Borough": "dropoff_borough",
            }
        ),
        on="dropoff_location_id",
        how="left",
    )

    df["borough_pair"] = (
        df["pickup_borough"]
        + "_"
        + df["dropoff_borough"]
    )

    df["same_borough"] = (
        df["pickup_borough"]
        == df["dropoff_borough"]
    ).astype(np.int8)

    return df

train_full = add_boroughs(train_full, zones)
test_full = add_boroughs(test_full, zones)
```

```python
def make_target_encoding(
    train,
    cols,
    target,
    feature_name,
    smoothing=50,
):
    global_mean = train[target].mean()

    stats = (
        train
        .groupby(cols)[target]
        .agg(["mean", "count"])
    )

    stats[feature_name] = (
        stats["mean"] * stats["count"]
        + global_mean * smoothing
    ) / (stats["count"] + smoothing)

    return (
        stats[[feature_name]]
        .reset_index(),
        global_mean,
    )

route_stats, global_mean = make_target_encoding(
    train_full,
    ["pickup_location_id", "dropoff_location_id"],
    "trip_duration_minutes",
    "route_avg_duration",
)

route_stats, global_mean = make_target_encoding(
    train_full,
    ["pickup_location_id", "dropoff_location_id"],
    "trip_duration_minutes",
    "route_avg_duration",
)

route_hour_stats, _ = make_target_encoding(
    train_full,
    [
        "pickup_location_id",
        "dropoff_location_id",
        "pickup_hour",
    ],
    "trip_duration_minutes",
    "route_hour_avg",
)

pickup_stats, _ = make_target_encoding(
    train_full,
    ["pickup_location_id"],
    "trip_duration_minutes",
    "pickup_zone_avg",
)

dropoff_stats, _ = make_target_encoding(
    train_full,
    ["dropoff_location_id"],
    "trip_duration_minutes",
    "dropoff_zone_avg",
)

hour_stats, _ = make_target_encoding(
    train_full,
    ["pickup_hour"],
    "trip_duration_minutes",
    "hour_avg_duration",
)

hour_day_stats, _ = make_target_encoding(
    train_full,
    ["pickup_hour", "pickup_day_of_week"],
    "trip_duration_minutes",
    "hour_day_avg",
)

def apply_encoding(
    df,
    stats,
    cols,
    feature,
    fill_value,
):
    df = df.merge(
        stats,
        on=cols,
        how="left",
    )

    df[feature] = df[feature].fillna(fill_value)

    return df
```

```python
train_full = apply_encoding(
    train_full,
    route_stats,
    ["pickup_location_id", "dropoff_location_id"],
    "route_avg_duration",
    global_mean,
)

test_full = apply_encoding(
    test_full,
    route_stats,
    ["pickup_location_id", "dropoff_location_id"],
    "route_avg_duration",
    global_mean,
)
```

```python
candidate_features = [
    "route_avg_duration",
    "route_hour_avg",
    "pickup_zone_avg",
    "dropoff_zone_avg",
    "hour_avg_duration",
    "hour_day_avg",
]

corrs = (
    test_full[candidate_features + ["trip_duration_minutes"]]
    .corr(numeric_only=True)["trip_duration_minutes"]
    .drop("trip_duration_minutes")
    .sort_values(ascending=False)
)

print(corrs)
```

```python
import pandas as pd
from lightgbm import LGBMRegressor
from xgboost import XGBRegressor

# ==========================================================
# FEATURES AVAILABLE IN BOTH TRAIN AND TEST
# ==========================================================

features = [
    "pickup_hour",
    "pickup_day_of_week",
    "pickup_month",
    "passenger_count",
    
    "pickup_location_id",
    "dropoff_location_id",

    "route_avg_duration",
    "route_frequency",

    "is_weekend",
    "rush_hour",
    "is_night",
    
    
    "same_borough",
]

target = "trip_duration_minutes"

X_train = train_full[features]
y_train = train_full[target]

# ==========================================================
# LIGHTGBM
# ==========================================================

model_lgb = LGBMRegressor(
    n_estimators=500,
    learning_rate=0.05,
    random_state=42,
    n_jobs=-1,
)

model_lgb.fit(X_train, y_train)

importance_lgb = (
    pd.Series(
        model_lgb.booster_.feature_importance(importance_type="gain"),
        index=features,
    )
    .sort_values(ascending=False)
)

print("\n========== LightGBM ==========")
print(importance_lgb)

# ==========================================================
# XGBOOST
# ==========================================================

model_xgb = XGBRegressor(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=8,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    tree_method="hist",
    n_jobs=-1,
)

model_xgb.fit(X_train, y_train)

importance_xgb = (
    pd.Series(
        model_xgb.feature_importances_,
        index=features,
    )
    .sort_values(ascending=False)
)

print("\n========== XGBoost ==========")
print(importance_xgb)
```

```python
test_full = apply_encoding(
    test_full,
    route_hour_stats,
    ["pickup_location_id","dropoff_location_id","pickup_hour"],
    "route_hour_avg",
    global_mean
)

test_full = apply_encoding(
    test_full,
    pickup_stats,
    ["pickup_location_id"],
    "pickup_zone_avg",
    global_mean
)

test_full = apply_encoding(
    test_full,
    dropoff_stats,
    ["dropoff_location_id"],
    "dropoff_zone_avg",
    global_mean
)

test_full = apply_encoding(
    test_full,
    hour_stats,
    ["pickup_hour"],
    "hour_avg_duration",
    global_mean
)

test_full = apply_encoding(
    test_full,
    hour_day_stats,
    ["pickup_hour","pickup_day_of_week"],
    "hour_day_avg",
    global_mean
)
```

```python
route_freq = (
    train_full
    .groupby(["pickup_location_id", "dropoff_location_id"])
    .size()
    .rename("route_frequency")
    .reset_index()
)

train_full = train_full.merge(
    route_freq,
    on=["pickup_location_id", "dropoff_location_id"],
    how="left",
)

test_full = test_full.merge(
    route_freq,
    on=["pickup_location_id", "dropoff_location_id"],
    how="left",
)

train_full["route_frequency"] = train_full["route_frequency"].fillna(0)
test_full["route_frequency"] = test_full["route_frequency"].fillna(0)
```

```python
import pandas as pd
from lightgbm import LGBMRegressor  # Assuming LightGBM based on `.booster_` syntax

# 1. Define your baseline features
baseline_l = [
    "pickup_hour",
    "pickup_day_of_week",
    "pickup_month",
    "passenger_count",
    "trip_distance",
    "pickup_location_id",
    "dropoff_location_id",
]

# 2. Split your data into features (X) and target (y)
# (Replacing 'trip_duration_minutes' with your actual target variable name)
X_train = train_full[baseline_l]
y_train = train_full["trip_duration_minutes"]

# 3. Initialize and fit your model
model = LGBMRegressor(random_state=42)
model.fit(X_train, y_train)

# 4. Your feature importance calculation (Now it works perfectly!)
importance = (
    pd.Series(
        model.booster_.feature_importance(importance_type="gain"),
        index=X_train.columns,
    )
    .sort_values(ascending=False)
)

print(importance)
```

```python
import pandas as pd
from xgboost import XGBRegressor

# 1. Define your baseline features
baseline_x = [
    "pickup_hour",
    "pickup_day_of_week",
    "pickup_month",
    "passenger_count",
    "trip_distance",
    "pickup_location_id",
    "dropoff_location_id",
]

# 2. Split your data into features (X) and target (y)
X_train = train_full[baseline_x]
y_train = train_full["trip_duration_minutes"]

# 3. Initialize and fit your XGBoost model
model = XGBRegressor(random_state=42)
model.fit(X_train, y_train)

# 4. XGBoost specific importance calculation (gain-based)
importance = (
    pd.Series(
        model.get_booster().get_score(importance_type="gain")
    )
    .sort_values(ascending=False)
)

print(importance)
```
